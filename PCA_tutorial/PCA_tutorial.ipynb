```python
import matplotlib.pyplot as plt
import numpy as np

from sklearn.decomposition import PCA
```

```python
rng = np.random.RandomState(0)
n_samples = 500
cov = [[3, 3], [3, 4]]
X = rng.multivariate_normal(mean=[0, 0], cov=cov, size=n_samples)
```

## One projection vector

```python
pca = PCA(n_components=1).fit(X) 
```

```python
Z = pca.components_@np.transpose(X) 
# You can also use `Z = scaler.fit_transform(X)`
```

```python
color = ["red", "blue"]
component_axes = ["1st", "2nd"]

plt.figure(figsize=(5,5))

plt.scatter(X[:, 0], X[:, 1], alpha=0.3, label="original samples $X$")
for i, (comp, var) in enumerate(zip(pca.components_, pca.explained_variance_)):
    comp = comp * var  # scale component by its variance explanation power 
    plt.arrow(0, 0, comp[0], comp[1], width=0.1, label="%s principal component axis" % component_axes[i],
        linewidth=2,
        color=color[i])

plt.gca().set(
    aspect="equal",
    title="Original data distribution vs principal component axes",
    xlabel="1st feature",
    ylabel="2nd feature",
)
plt.legend( ) 

plt.tight_layout()
plt.savefig("PCA-axis_one_axis.png")
```

```python
# Result after projection.   
Z_tr = np.transpose(Z)
Z_x = Z_tr[:, 0]
Z_y = np.zeros_like(Z_x)

plt.figure(figsize=(8,2))
plt.scatter(Z_x, Z_y, alpha=0.3, label="projected samples $Z$", color="grey") 
plt.gca().set( 
    title="Result of the projection on PCA axes",  
)
plt.xlabel("1st principal component axis", color="red") 
plt.legend() 
plt.tight_layout()
plt.savefig("PCA-projection-one-axis.png")
```

## Reverse the projection

```python
X_approx_pca = np.transpose(pca.components_)@Z 
X_approx_psue = np.linalg.pinv(pca.components_)@Z 
```

```python
plt.figure(figsize=(5,5))
plt.scatter(X[:,0], X[:,1], alpha=0.5, label="Original data", color="dodgerblue") 
plt.scatter(X_approx_pca[1,:], X_approx_pca[0,:], s=20,  alpha=0.8, label="Transpose $Q^T_{p=1}$", color="red") 
plt.scatter(X_approx_psue[1,:], X_approx_psue[0,:], s=10, alpha=0.5, label="Pseudo inverse $Q^T_{p=1}$", color="lime") 
plt.gca().set(
    aspect="equal",
    title="Original data vs 1-dimensional approximation",
    xlabel="1st feature",
    ylabel="2nd feature",
)
plt.legend( ) 
plt.savefig("PCA-reverse.png")
```
